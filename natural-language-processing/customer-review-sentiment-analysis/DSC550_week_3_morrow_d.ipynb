#### Exercise 1.b Use TextBlob to classify each movie review as positive or negative. Assume that a polarity score greater than or equal to zero is a positive sentiment and less than 0 is a negative sentiment.

In this step, I will use the TextBlob library to analyze the sentiment of each movie review. 
TextBlob provides a polarity score ranging from -1 to 1, where negative values indicate negative sentiment and positive values indicate positive sentiment.



```python
from textblob import TextBlob

#Create function to classify sentiment with TextBlob
def get_sentiment(review):
    
    #Get polarity score
    polarity = TextBlob(review).sentiment.polarity
    
    #Apply classification rule
    if polarity >= 0:
        return 1
    else:
        return 0

#Apply function to all reviews
df["predicted_sentiment"] = df["review"].apply(get_sentiment)

#Output first 5 rows 
df[["sentiment", "predicted_sentiment"]].head()
```

#### Exercise 1.b explanation:

The TextBlob sentiment analyzer was applied to each review, and then the predicted sentiment was stored in a new column called `predicted_sentiment`. 
Comparing the actual and predicted values in the sample output shows that TextBlob correctly identifies some reviews and misclassifies others.

Similar to the first exercise using TextBlob, I will now use the VADER (Valence Aware Dictionary and sEntiment Reasoner) sentiment analyzer to classify each movie review. 

The compound score from VADER ranges from -1 to 1. For this assignment, I will classify reviews as positive if the compound score is greater than or equal to 0, and negative if it is less than 0.


#### Extra Credit 1 Use Vader to classify each movie review as positive or negative. Assume that a polarity score greater than or equal to zero is a positive sentiment and less than 0 is a negative sentiment.

In this step, I will use the Vader library to analyze the sentiment of each movie review. Vader, similar to TextBlob, provides a polarity score ranging from -1 to 1, where negative values indicate negative sentiment and positive values indicate positive sentiment.

```python
from nltk.sentiment import SentimentIntensityAnalyzer

#Initialize VADER
sia = SentimentIntensityAnalyzer()

#Function to classify sentiment
def get_vader_sentiment(review):
    score = sia.polarity_scores(review)
    
    if score['compound'] >= 0:
        return 1
    else:
        return 0

#Apply function
df["vader_sentiment"] = df["review"].apply(get_vader_sentiment)

#Ouput first 5 rows
df[["sentiment", "vader_sentiment"]].head()
```

```python
vader_accuracy = (df["sentiment"] == df["vader_sentiment"]).mean()
vader_accuracy
```

#### Extra Credit 1 explanation:
The VADER sentiment analyzer achieved an accuracy of approximately 0.694,  which is slightly higher than the TextBlob accuracy of 0.685. Both models performed better than random guessing (0.50), indicating that they are both effective at capturing sentiment in the movie reviews.

While the difference in accuracy is relatively small, VADER appears to perform slightly better on this dataset. This may be because VADER is specifically designed for sentiment analysis and is optimized for interpreting text-based data (according to the reading and information online). Overall, both models provide useful results, but VADER demonstrates a slight advantage in this case.

#### Part 2 Prepping text for a custom model
This is using the same dataset from the above exercise. 

#### Exercise 2.a Convert all text to lowercase letters
This is done to ensure all text is treated the same, the information will be placed in a new column titled "review_lower".

```python
#Convert review text to lowercase and store in new column
df["review_lower"] = df["review"].str.lower()

#Output first few rows
df["review_lower"].head()
```

#### Exercise 2.b Remove punctuation and special characters from the text.
This involved created a function with clean_text, and adding a new column "review_clean" to store the updated data. 

```python
import re

#Function to remove punctuation and special characters
def clean_text(text):
    
    #Replace anything not a letter or space with a space
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    
    return text

#Apply Clean function to lowercase text
df["review_clean"] = df["review_lower"].apply(clean_text)

#Output first few rows
df["review_clean"].head()
```

#### Exercise 2.c Remove stop words.
Stop words are common words such as "the", "and", and "is" that do not add meaningful value to the analysis. 
This will need the NLKT package for stopwords to be identified. 
After the stopwords are removed the data will be stored in a new column called "review_no_stopwords".

```python
from nltk.corpus import stopwords

#Load English stopwords
stop_words = set(stopwords.words('english'))

#Function to remove stopwords
def remove_stopwords(text):
    
    #Split text into words
    words = text.split()
    
    #Keep only words not in stopwords list
    filtered_words = [word for word in words if word not in stop_words]
    
    #Bring words back into single string
    return " ".join(filtered_words)

#Apply function to cleaned text
df["review_no_stopwords"] = df["review_clean"].apply(remove_stopwords)

#Output first few rows
df["review_no_stopwords"].head()
```

#### Exercise 2.d Apply NLTK’s PorterStemmer.
Stemming data reduces word to their root form, which will help in grouping data. 
The Stemmed data will be put in a new column called "review_stemmed".
The output in this column will look odd if you are not familiar with the PorterStemmer. 

```python
from nltk.stem import PorterStemmer

#Create PorterStemmer object
stemmer = PorterStemmer()

#Function to apply stemming to each review
def stem_text(text):
    
    #Split review into individual words
    words = text.split()
    
    #Apply stemming to each word
    stemmed_words = [stemmer.stem(word) for word in words]
    
    #Bring stemmed words back into single string
    return " ".join(stemmed_words)

#Apply stemming to the stopword-removed text
df["review_stemmed"] = df["review_no_stopwords"].apply(stem_text)

#Output the first few rows
df["review_stemmed"].head()
```

#### Exercise 2.e * Create a bag-of-words matrix from your stemmed text (output from (4)), where each row is a word-count vector for a single movie review (see sections 5.3 & 6.8 in the Machine Learning with Python Cookbook).

A bag-of-words matrix represents each review as a vector of word counts, 
where each column corresponds to a unique word in the dataset. This will allow the text data to be converted to a numerical format. Each value in the matrix will represent the number of times a word appears in a review.

```python
from sklearn.feature_extraction.text import CountVectorizer

#Initialize CountVectorizer
vectorizer = CountVectorizer()

#Transform stemmed text into bag-of-words matrix
bow_matrix = vectorizer.fit_transform(df["review_stemmed"])

#Output shape of matrix
bow_matrix.shape
```

#### Exercise 2.f Display the dimensions of your bag-of-words matrix. The number of rows in this matrix should be the same as the number of rows in your original data frame.
For this data, each row represents a movie review and each column represents a unique word in the dataset. The dimensions from the above are 25,000 Rows, 49,642 Columns - Unique words. Therefor the 25,000 rows matches the original dataset. 


#### Exercise 2.g Create a term frequency-inverse document frequency (tf-idf) matrix from your stemmed text, for your movie reviews (see section 6.9 in the Machine Learning with Python Cookbook).

Lastly, creating a TF-IDF (term frequency–inverse document frequency) matrix from the stemmed. Unlike the bag-of-words matrix, TF-IDF reflects the importance of each word by reducing the weight of common words and increasing the importance of less frequent words.



```python
from sklearn.feature_extraction.text import TfidfVectorizer

#Initialize TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer()

#Transform stemmed text
tfidf_matrix = tfidf_vectorizer.fit_transform(df["review_stemmed"])

#Output shape of matrix
tfidf_matrix.shape
```

The TF-IDF matrix has the same shape as the bag-of-words matrix, with 25,000 rows and 49,642 columns. Each row representing a movie review, and each column representing a unique word.
